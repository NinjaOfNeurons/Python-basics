### What's the difference between transform() and agg() in a groupby context?

"agg() reduces each group to one or more summary values, so the output is usually smaller than the original DataFrame. transform() performs a group-level calculation but returns a result with the same number of rows as the original data, broadcasting the group result back to each row. I use transform() when I need group statistics alongside the original rows."

````python
df.groupby("Department")["Salary"].agg("mean")
````

````python
df["Dept_Avg"] = df.groupby("Department")["Salary"].transform("mean")
````

How do you pivot a DataFrame using pivot_table() vs. pivot() — what's the key difference?


Both reshape a DataFrame from long format → wide format, but the key difference is:

pivot() requires unique combinations, while pivot_table() can handle duplicates by aggregating them.


| Department | City   | Salary |
| ---------- | ------ | -----: |
| IT         | Delhi  |     50 |
| IT         | Mumbai |     70 |
| HR         | Delhi  |     40 |
| HR         | Mumbai |     50 |



df.pivot(
    index="Department",
    columns="City",
    values="Salary"
)


| Department | Delhi | Mumbai |
| ---------- | ----: | -----: |
| HR         |    40 |     50 |
| IT         |    50 |     70 |

"pivot() reshapes the DataFrame but requires each index-column combination to be unique. If duplicate combinations exist, it raises an error. pivot_table() is more flexible because it can handle duplicates using an aggregation function such as mean, sum, or count. So I use pivot() for straightforward reshaping with unique combinations, and pivot_table() when aggregation may be required."
